```python
# HR/BHR raw text samples

employee_feedback = "OMG!!! 😓 The onboarding process was REALLY confusing!!! But the HR team was super supportive. #EmployeeExperience"

manager_feedback = "Employees are learning new leadership skills, but workload pressure is increasing rapidly."

candidate_review = "The interview process was smooth 😊, however communication delays from recruitment were frustrating!!!"
```

TOKENISATION

This breaks HR text into:
sentences
words

```python
from nltk.tokenize import word_tokenize, sent_tokenize
import nltk

samples = {
    "Employee Feedback": employee_feedback,
    "Manager Feedback": manager_feedback,
    "Candidate Review": candidate_review
}

for name, text in samples.items():

    print(f"\n{name}")
    print("-" * 50)

    # Ensure punkt_tab is downloaded right before use
    nltk.download('punkt_tab', quiet=True)

    # Sentence Tokenization
    sentences = sent_tokenize(text)

    print("Sentence Tokens:")
    print(sentences)

    # Word Tokenization
    words = word_tokenize(text)

    print("\nWord Tokens:")
    print(words)
```

AI separates:

*   onboarding

*   Supportive
*   workload
*   Leadership
*   Recruitment
These becomes analyzable HR keywords.



workload
leadership
recruitment
These become analyzable HR keywords.

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

for name, text in samples.items():

    words = word_tokenize(text)

    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]

    print(f"\n{name}")
    print("-" * 50)

    print("Before Stopword Removal:")
    print(words)

    print("\nAfter Stopword Removal:")
    print(filtered_words)
```

Words like:
the
was
but
are
get removed.
AI focuses on important HR signals: ✅ onboarding
✅ workload
✅ leadership
✅ recruitment
✅ supportive

Regex Cleaning
This removes:
emojis
punctuation
hashtags

```python
import re

for name, text in samples.items():

    # Remove punctuation, emojis and symbols
    cleaned_text = re.sub(r'[^A-Za-z\s]', '', text)

    print(f"\n{name}")
    print("-" * 50)

    print("Before Cleaning:")
    print(text)

    print("\nAfter Cleaning:")
    print(cleaned_text)
```

Stemming & Lemmatization

```python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

hr_words = [
    "learning",
    "employees",
    "recruitment",
    "working",
    "leaders"
]

print("Word | Stemmed | Lemmatized")
print("-" * 50)

for word in hr_words:

    stemmed = stemmer.stem(word)

    lemmatized = lemmatizer.lemmatize(word)

    print(f"{word} | {stemmed} | {lemmatized}")
```

AI understands:
learning → learn
employees → employee
working → work
Now HR systems can group similar employee themes together.

## Why Text Preprocessing Matters in HR

HR systems receive large amounts of unstructured text from:
- employee surveys
- engagement feedback
- recruitment reviews
- performance comments

Text preprocessing helps AI clean and structure this data before analysis.

### Tokenization
Breaks feedback into sentences and words.

### Stopword Removal
Removes unnecessary common words so AI focuses on meaningful HR signals.

### Regex Cleaning
Removes punctuation, emojis, hashtags and symbols to improve processing consistency.

### Stemming
Reduces words to root forms for better grouping of employee concerns.

### Lemmatization
Converts words into proper dictionary forms to improve semantic understanding.

Without preprocessing:
- HR analytics become inaccurate
- sentiment analysis becomes noisy
- employee themes are fragmented
- AI models struggle to identify patterns

Preprocessing improves:
- employee sentiment analysis
- workforce analytics
- recruitment insights
- engagement trend detection
