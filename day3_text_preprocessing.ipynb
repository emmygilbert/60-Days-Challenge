```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """In a growing organization, the Business Human Resources (BHR) team is responsible for aligning HR strategies with business objectives to drive performance and growth.
The team collaborates with leadership to identify hiring needs, manage recruitment, and ensure employee satisfaction.
They analyze attrition trends and performance data to improve productivity and retention."""
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words('english'))

clean_tokens = [
    word.lower() for word in tokens
        if word.lower() not in stop_words and word not in string.punctuation
        ]

print(clean_tokens)
```

```python
from collections import Counter

bow = Counter(clean_tokens)
print(bow)
```

```python
print("Original:", text)
print("Processed:", " ".join(clean_tokens))
```
