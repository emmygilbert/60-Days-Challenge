```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "We need to hire software engineers for the new project",
    "Recruitment is required for technical roles in the organization",
    "Employee engagement is low and needs improvement",
    "Staff are not satisfied with the current work environment",
    "The company is planning to expand into international markets"
]

for i, s in enumerate(sentences):
    print(f"{i}: {s}")
```

```python
embeddings = model.encode(sentences)

print("Embedding Shape:", len(embeddings), "x", len(embeddings[0]))
```

```python
similarity_matrix = cosine_similarity(embeddings)

df = pd.DataFrame(similarity_matrix, index=sentences, columns=sentences)
df
```

```python
print("📊 HR Similarity Analysis:\n")

for i in range(len(sentences)):
    for j in range(i+1, len(sentences)):
        score = cosine_similarity([embeddings[i]], [embeddings[j]])[0][0]
        print(f"{i} ↔ {j} = {score:.2f}")
```

```python
print("\nf50d Key HR Insights:\n")

print("Recruitment similarity:",
      cosine_similarity([embeddings[0]], [embeddings[1]])[0][0])

print("Employee satisfaction similarity:",
      cosine_similarity([embeddings[2]], [embeddings[3]])[0][0])

print("HR vs Business strategy:",
      cosine_similarity([embeddings[0]], [embeddings[4]])[0][0])
```

🧠 Reflection

In this experiment, I applied sentence embeddings to real HR scenarios such as recruitment needs, employee engagement, and business strategy.

I observed that:
- Recruitment-related sentences showed high similarity scores, even with different wording.
- Employee satisfaction statements also had strong similarity, indicating shared meaning.
- Business strategy sentences had low similarity with HR-focused content.

This demonstrates how embeddings can help HR teams:
- Match job descriptions with candidate profiles
- Analyze employee feedback automatically
- Group similar HR issues for better decision-making

Overall, embeddings provide a powerful way to bring intelligence into HR analytics and decision-making.
