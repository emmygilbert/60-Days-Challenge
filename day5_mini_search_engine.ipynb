```python
sentences = [
    "Employee engagement improves productivity.",
    "Training programs help employees grow.",
    "Remote work increases flexibility.",
    "Recognition boosts employee morale.",
    "Work-life balance reduces burnout.",
    "Performance reviews help career development.",
    "Leadership training improves management skills.",
    "Flexible schedules improve satisfaction.",
    "Mental health support is important at work.",
    "Good communication builds strong teams.",
    "Employee retention reduces hiring costs.",
    "Diversity improves workplace culture.",
    "HR analytics helps decision making.",
    "Learning opportunities attract talent.",
    "Feedback improves employee performance.",
    "Rewards increase motivation.",
    "Conflict resolution improves teamwork.",
    "Onboarding helps new employees settle quickly.",
    "Career growth increases retention.",
    "Positive culture attracts skilled employees."
]
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

embeddings = model.encode(sentences)

print("Embeddings created successfully!")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def semantic_search(query):
    query_embedding = model.encode([query])
    similarities = cosine_similarity(query_embedding, embeddings)[0]
    top_indices = np.argsort(similarities)[::-1][:3]
    print(f"\nQuery: {query}\n")
    print("Top 3 Similar Results:\n")
    for idx in top_indices:
        print(f"- {sentences[idx]} (Score: {similarities[idx]:.2f})")
```

```python
semantic_search("How can companies improve employee happiness?")
```

```python
query = "employee happiness"

print("Keyword Search Results:\n")

for sentence in sentences:
    if "happiness" in sentence.lower():
            print("-", sentence)

            print("\nSemantic Search Results:")
            semantic_search(query)
```
